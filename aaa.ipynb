```python
# 삼항 연산자를 이용한 함수 예제

def max_value(a, b):
    """두 수 중 큰 값을 반환"""
    return a if a > b else b

def is_even(num):
    """짝수 여부를 판별"""
    return "짝수" if num % 2 == 0 else "홀수"

def grade(score):
    """점수에 따른 학점 반환"""
    return "합격" if score >= 60 else "불합격"

# 테스트
print(max_value(10, 20))
print(is_even(7))
print(grade(75))
```

```python
def multiplication_table(n):
    """n단의 구구단을 출력"""
    for i in range(1, 10):
        print(f"{n} × {i} = {n * i}")

# 테스트
multiplication_table(3)
```

```python
def all_multiplication_tables():
    """2단부터 9단까지의 구구단을 모두 출력"""
    for n in range(2, 10):
        multiplication_table(n)
        print()  # 각 단 사이에 빈 줄 추가

# 테스트
all_multiplication_tables()
```
